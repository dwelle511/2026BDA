```python
#pd.pivot_table() : create a spreadsheet-style pivot table as a DataFrame
#
import pandas as pd
import numpy as np
pivot_df=pd.DataFrame(
    {
        "a":["foo","foo","foo","foo","foo","bar","bar","bar","bar"],
        "b":["one","one","one","two","two","one","one","two","two"],
        "c":["small","large","large","small","small","large","smnall","small","large"],
        "d":[1,2,2,3,3,4,5,6,7],
        "e":[2,4,5,5,6,6,8,9,9],

    }
)
pivot_df
```

```python
#the first example aggregates values by talking the sum
# DataFrame.pivot() ; pivot without aggregation that can handle non-numeric data
table=pd.pivot_table(
    pivot_df,values="d",index=["a","b"],columns=["c"],aggfunc="sum"
)
table
# 설명 : 멀티 인티인텍스 index["a","b"]
```

```python
table=pd.pivot_table(
  pivot_df,values="d",index=["a","b"],columns=["c"],aggfunc="sum",fill_value=0
)
table
```

```python
table=pd.pivot_table(
    pivot_df, values=["d","e"],index=["a","c"],aggfunc={"d":"mean","e":"mean"}
)
table
```

```python
table=pd.pivot_table(
    pivot_df,values=["d","e"],index=["a","c"],aggfunc={"d":"mean","e":["min","max","mean"]}
)
table
```

```python
#Categoricals
#pandas can include categorical data in a DataFrame. For full docs, see the categorical
#introductuiob and the APIdocumentation

df=pd.DataFrame(
    {"id":[1,2,3,4,5,6],"raw_grade":["a","b","b","a","a","e"]}
)
```

```python
#converting the raw grades to a categorical data type
df["grade"]=df["raw_grade"].astype("category")
df["grade"]
```

```python
#rename the categories to more meaningful names
new_categories=["very good","good","very bad"]
df["grade"]=df["grade"].cat.rename_categories(new_categories)
df["grade"]
```

```python
#reorder the categories and simultaneously add the missing categories(method under Series.cat())
#return a new Series bt default()
df["grade"]=df["grade"].cat.set_categories(
    ["very bad","bad","medium","good","very good"]
)
df["grade"]
```

```python
#sorting si per order in the categories, not lexical order
df.sort_values(by="grade")
```
