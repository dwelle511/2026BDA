### 수정 불가능한 immutable 객체 &함수 id()

- 파이썬에서 객체의 고유 식별자(unique identifier)을 이햐하는데 중요한 역할
- immutable 객체는 변경 불가
  + 이런 객체는 생성 후 변경 불가
  + 변경하렴녀 새로운 객체 생성 = id() value 달라짐
- 함수 id()
  + 모든 객체의 고유 식별자를 반화
  + id() 함수 사용 = 두 객체가 동일한 메모리 위치를 가리키는지 여부를 확인 가능
